# ❖ Task 1 — Data Import & Setup

```python
import pandas as pd

# import data from github
url = "https://raw.githubusercontent.com/GeethaGunasekaran1/Dataset_rep/main/social_media_engagement_5000.csv"

# Load dataset
df = pd.read_csv(url)
```

```python
df
```

```python
# taking copy for the data
df_safe = df.copy()
```

```python
#Check/convert data types
print(df.dtypes)
```

```python
# changeing data types
# changeing user id into object
df['user_id'] = df['user_id'].astype('object')

# changeing user id into object
df['post_id'] = df['post_id'].astype('object')

# changing data types of posted at into date



df['posted_at'] = pd.to_datetime(df['posted_at'],format="mixed",errors='coerce',dayfirst='true')
```

# ❖ Task 2 — Data Cleaning

```python
# checkinf for duplicated value
df.duplicated().sum()
```

```python
# checking for null values
df.isnull().sum()
```

# age coloumn

```python
# find describtion of age
df["age"].describe()
```

```python
# imputed age values
df['age'].fillna(df['age'].median(), inplace=True)
```

```python
# changeing data types of age
df['age'] = df['age'].astype('int')
```

# Gender Coloumn

```python
#FIND UNIQUES VALUES
df['gender'].unique()
```

```python
#FIND VALUES COUNT
df['gender'].value_counts()
```

```python
df["gender"] = df['gender'].fillna("Male")
```

```python
#POST TYPES COLOUMN
df['post_type'] = df['post_type'].str.title().str.strip()
```

```python
#POST CATEGORY COLOUMN
df['post_category'] = df['post_category'].str.title().str.strip()
```

# Likes coloumn

```python
#inputing likes coloumn
df["likes"]=df["likes"].fillna(df['likes'].median())
```

```python
#changing data types
df['likes'] = df['likes']. astype('int')
```

# comments coloumn

```python
#IMPUTING COLUMN
df['comments'].fillna(df['comments'].median(), inplace=True)
```

```python
#changeing data types
df['comments'] = df['comments'].astype('int')
```

# shares coloumn

```python
#inputing coloumn
df["shares"] = df["shares"].fillna(df["shares"].median(),inplace=True)
```

```python
df["shares"] = df["shares"].fillna(0)
df["shares"] = df["shares"].astype(int)
```

```python
#device column
df['device_type'] = df['device_type'].str.title().str.strip()
```

# sentiment coloumn

```python
#FIND UNIQUES VALUES
df['sentiment'].unique()
```

```python
#finding values count
df['sentiment'].value_counts()
```

```python
#imputing value
df["sentiment"] = df['sentiment'].fillna("positive")
```

```python
#formating the column
df['sentiment'] = df['sentiment'].str.capitalize().str.strip()
```

```python
#counting hashtags

df['hashtag_count'] = df['hashtags'].str.count(r'#\w+')
```

# ❖ Task 5 — Statistical Analysis

```python
# Select relevant columns
cols = ['likes', 'comments', 'shares', 'watch_time_sec', 'engagement_rate', 'follower_count']

data = df[cols]

# Mean, Median, Mode
print("Mean:\n", data.mean())
print("\nMedian:\n", data.median())
print("\nMode:\n", data.mode().iloc[0])

# Standard Deviation & Variance
print("\nStandard Deviation:\n", data.std())
print("\nVariance:\n", data.var())

# Percentiles (25th, 50th, 75th)
print("\n25th Percentile:\n", data.quantile(0.25))
print("\n50th Percentile:\n", data.quantile(0.50))  # same as median
print("\n75th Percentile:\n", data.quantile(0.75))

# (Skewness & Kurtosis
print("\nSkewness:\n", data.skew())
print("\nKurtosis:\n", data.kurt())
```

# ❖ Task 6 — Data Visualization

# ● Matplotlib

```python
import matplotlib.pyplot as plt

# Scatter: likes vs impressions
plt.scatter(df['impression_count'], df['likes'], alpha=0.5, color='orange')
plt.xlabel('Impressions')
plt.ylabel('Likes')
plt.title('Likes vs Impressions')
plt.show()

```

```python
# Line: daily engagement trend
plt.plot(df['posted_at'], df['engagement_rate'], marker='o', linestyle='-')
plt.xticks(rotation=45)
plt.xlabel('Date')
plt.ylabel('Engagement Rate')
plt.title('Daily Engagement Trend')
plt.show()
```

```python
# Bar: posts by category
df['post_category'].value_counts().plot(kind='bar', color='green')
plt.title('Posts by Category')
plt.show()
```

```python
# Pie: gender distribution
df['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.show()
```

```python
# Histogram: age
plt.hist(df['age'], bins=15, color='purple', edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution')
plt.show()
```

```python
# Box: engagement rate
plt.boxplot(df['engagement_rate'])
plt.title('Engagement Rate Distribution')
plt.show()
```

# ● Seaborn

```python
import seaborn as sns
import numpy as np

# Count plot: post type
sns.countplot(x='post_type', data=df)
plt.title('Post Type Count')
plt.show()
```

```python
# Bar plot: avg likes by category
sns.barplot(x='post_category', y='likes', data=df, estimator=np.mean)
plt.title('Average Likes by Category')
plt.show()
```

```python
# Violin: followers vs sentiment
sns.violinplot(x='sentiment', y='follower_count', data=df)
plt.title('Followers vs Sentiment')
plt.show()
```

```python
# Pair plot: numeric features
sns.pairplot(df[['likes','comments','shares','impression_count','engagement_rate']])
plt.show()
```

```python
# Heatmap: correlation matrix
corr = df[['likes','comments','shares','impression_count','engagement_rate']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Swarm plot: engagement vs device
sns.stripplot(x='device_type', y='engagement_rate', data=df, jitter=True)

plt.title('Engagement Rate vs Device Type')
plt.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
# --- Line Chart: Engagement Trend ---

line_fig = px.line(df, x='posted_at', y='engagement_rate',
                   title='Engagement Trend', markers=True)
line_fig.update_xaxes(rangeslider_visible=True)

# --- Bar Chart: Avg Likes by Category ---
bar_data = df.groupby('post_category')['likes'].mean().reset_index().sort_values('likes', ascending=False)
bar_fig = px.bar(bar_data, x='post_category', y='likes',
                 title='Avg Likes by Category')
bar_fig.update_traces(marker_color='teal')

# --- Scatter/Bubble Chart: Likes vs Impressions ---
scatter_fig = px.scatter(df, x='impression_count', y='likes',
                         size='follower_count', color='post_category',
                         hover_name='post_type',
                         title='Likes vs Impressions Bubble Chart',
                         trendline='ols')
scatter_fig.update_xaxes(type='log')

# --- Combine into Subplots ---
fig = make_subplots(rows=3, cols=1,
                    subplot_titles=("Engagement Trend",
                                    "Avg Likes by Category",
                                    "Likes vs Impressions Bubble Chart"),
                    vertical_spacing=0.15)

# Add traces from each fig
for trace in line_fig.data:
    fig.add_trace(trace, row=1, col=1)

for trace in bar_fig.data:
    fig.add_trace(trace, row=2, col=1)

for trace in scatter_fig.data:
    fig.add_trace(trace, row=3, col=1)

# Update layout
fig.update_layout(height=1200, width=900,
                  title_text="Interactive Engagement Dashboard",
                  showlegend=True)

fig.show()
```
